# Data Parsing using LangChain Loaders

In LangChain, we can use Document loaders, they load different files like CSV, TXT, PDF, DOCX, HTML, and JSON. Internally, they may use parser libraries like pypdf, docx2txt, unstructured, or jq to extract the content. Then LangChain converts that extracted content into Document(page_content, metadata) format for the RAG pipelines

```python
from langchain_community.document_loaders import (
    TextLoader,
    PyPDFLoader,
    CSVLoader,
    Docx2txtLoader,
    UnstructuredHTMLLoader,
    UnstructuredMarkdownLoader,
    JSONLoader,
    WebBaseLoader
)
```

## PDF Loader

```python
loader = PyPDFLoader("data/sample_document.pdf")
```

```python
docs = loader.load()
docs
```

```python
docs[0].metadata
```

```python
print(docs[0].page_content[:500])
```

## Text Loader

```python
text_loader = TextLoader("data/sample_text.txt")
```

```python
text_docs = text_loader.load()
text_docs
```

```python
text_docs[0].metadata
```

```python
print(text_docs[0].page_content[:500])
```

## CSV Loader

```python
csv_loader = CSVLoader("data/sample_data.csv")
csv_docs = csv_loader.load()
csv_docs
```

```python
csv_docs[0].metadata
```

```python
csv_docs[0].page_content
```

## JSON Loader

```python
json_loader = JSONLoader(file_path="data/sample_data.json", jq_schema=".[]", text_content=False)
json_docs = json_loader.load()
json_docs
```

```python
json_docs[0].metadata
```

```python
json_docs[0].page_content
```

## HTML Loader

```python
html_loader = UnstructuredHTMLLoader("data/sample_page.html")
html_docs = html_loader.load()
html_docs
```

```python
print(html_docs[0].page_content)
```

## DocX Loader

```python
docx_loader = Docx2txtLoader("data/sample_document.docx")
docx_docs = docx_loader.load()
docx_docs
```

```python
print(docx_docs[0].page_content)
```

## MarkDown Loader

```python
markdown_loader = UnstructuredMarkdownLoader("data/sample.md")
markdown_docs = markdown_loader.load()
markdown_docs
```

```python
print(markdown_docs[0].page_content)
```

## Web Base Loader

```python
web_loader = WebBaseLoader("https://www.geeksforgeeks.org/dsa/dsa-tutorial-learn-data-structures-and-algorithms/")
web_docs = web_loader.load()
web_docs
```

```python
print(web_docs[0].page_content)
```
