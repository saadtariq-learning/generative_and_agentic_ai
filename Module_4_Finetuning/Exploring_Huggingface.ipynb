## Custom Dataset

```python
import pandas as pd
from datasets import Dataset
```

```python
data = {
    "text": [
        "I love this product! Works great ",
        "Terrible experience, I want a refund.",
        "Fast delivery and excellent packaging.",
        "Worst app I’ve ever used. Crashed in 1 minute.",
        "Super helpful support team, thanks a lot!",
        "Nothing worked. Waste of time."
    ],
    "label": [1, 0, 1, 0, 1, 0]  # 1 = Positive, 0 = Negative
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
dataset = Dataset.from_pandas(df)
dataset
```

### Pushing Dataset to Hub

```python
os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN_WRITE")
# notebook_login()
```

```python
dataset.push_to_hub("saadtariq/mycustomdataset")
```

```python
data = {
    "instruction": [
        "Translate English to Spanish",
        "Summarize the text",
        "Explain Machine Learning"
    ],
    "input": [
        "I love Machine Learning",
        "Artificial Intelligence is transforming industries.",
        ""
    ],
    "output": [
        "Me encanta el aprendizaje automático",
        "Artificial intelligence is changing how the world works.",
        "Machine Learning is a subset of AI."
    ]
}

new_dataset = Dataset.from_dict(data)

new_dataset.push_to_hub("saadtariq/mynewcustomdataset")
```
